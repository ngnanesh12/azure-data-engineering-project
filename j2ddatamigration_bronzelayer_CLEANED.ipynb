```python
null_checkist=['BranchId',
 'BranchArea',
 'productcode',
 'Category',
 'Category_Code',
 'Quantity',
 'salesdate',
 ]
```

```python
from pyspark.sql.functions import col, sum

null_counts_row = (
    new_df
    .select([col(c).isNull().alias(c) for c in null_checkist])
    .agg(*[sum(col(c).cast("int")).alias(c) for c in null_checkist])
    .collect()[0]
)
null_counts = {c: null_counts_row[c] for c in null_checkist}
display(null_counts)
if any(null_counts[c] > 0 for c in null_checkist):
    dbutils.notebook.exit("Null is there")
```

```python
from pyspark.sql.functions import col, count

duplicate_invoice_ids = (
    new_df.groupBy("InvoiceId")
    .agg(count("*").alias("duplicate_count"))
    .filter(col("duplicate_count") > 1)
)

display(duplicate_invoice_ids)

if duplicate_invoice_ids.count() > 0:
    dbutils.notebook.exit("Duplicate InvoiceId values found")
```
